### 데이터베이스(스키마)와 테이블 만들기

```python
import pymysql

conn = pymysql.connect(
    host='127.0.0.1',
    user='root',
    password='qwer1234'
)
cur = conn.cursor()
cur.execute("CREATE DATABASE IF NOT EXISTS testdb")
conn.close()
```

```python
conn = pymysql.connect(
    host='127.0.0.1',
    user='root',
    password='qwer1234',
    database='testdb'
)
cur = conn.cursor()
cur.execute("""
CREATE TABLE members 
(id INT AUTO_INCREMENT PRIMARY KEY,
 name VARCHAR(255), 
 address VARCHAR(255),
 email VARCHAR(255),
 phone_number VARCHAR(255),
 job VARCHAR(255),
 company VARCHAR(255)
 )
 """)
conn.commit()
print("테이블이 생성되었습니다.")
conn.close()
```

```python
from faker import Faker
fake = Faker('ko-KR')
print(fake.name())
print(fake.address())
print(fake.email())
print(fake.phone_number())
print(fake.job())
print(fake.company())
```

### CRUD - INSERT
- 데이터 추가

```python
cur.execute("""INSERT INTO members (name, address, email, phone_number, job, company)
 VALUES 
 (%s, %s, %s, %s, %s, %s)""",
 (fake.name(), fake.address(), 
 fake.email(), fake.phone_number(),
  fake.job(), fake.company())
  )
conn.commit()
print("데이터가 삽입되었습니다.")
```

```python
for i in range(10):
    cur.execute("""INSERT INTO members (name, address, email, phone_number, job, company)
     VALUES
        (%s, %s, %s, %s, %s, %s)""",
        (fake.name(), fake.address(),
         fake.email(), fake.phone_number(),
         fake.job(), fake.company())
    )
conn.commit()
```

```python
# executemany : INSERT문 하나에 데이터 목록을 통째로 넘기면, 반복문 없이 여러 행을 한 번에 저장 가능
members = [fake.name(), fake.address(), fake.email(), fake.phone_number(), fake.job(), fake.company()]
members

cur.execute("""
 INSERT INTO members (name, address, email, phone_number, job, company)
 VALUES (%s, %s, %s, %s, %s, %s)
 """,(fake.name(), fake.address(), fake.email(), fake.phone_number(), fake.job(), fake.company()))
conn.commit()

print("데이터가 삽입되었습니다.")
```

### CRUD - SELECT
- 데이터 조회

```python
cur.execute("SELECT * FROM members")
rows = cur.fetchall() # 전체조회
for row in rows:
    print(row)
```

```python
cur.execute("SELECT * FROM members WHERE ID = 1")
cur.fetchone() # 결과 조건이 1개
```

### CRUD - UPDATE
- 데이터 수정

```python
cur.execute("UPDATE members SET address = %s WHERE id = %s", ("서울시 강남구 테헤란로 123", 1))
conn.commit()
```

### CRUD - DELETE
- 데이터 삭제

```python
cur.execute("DELETE FROM members WHERE id = %s", (1,))
conn.commit()
```
