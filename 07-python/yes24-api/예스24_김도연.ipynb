- **항목**: 순위, 제목, 저자, 가격, 상세 정보(판매지수)

```python
from urllib.robotparser import RobotFileParser

rp = RobotFileParser()
rp.set_url("https://www.yes24.com/product/category/daybestselle") # robots.txt 주소
rp.read()
rp.can_fetch("*", "https://www.yes24.com/product/category/daybestselle") # ("User-Agent", "확인할 URL") - 크롤링 가능하면 True, 금지면 False
```

```python
import requests

url = "https://www.yes24.com/product/category/daybestselle"
response = requests.get(url)

response.encoding = response.apparent_encoding # 실제 내용을 보고 다시 추측해서 인코딩
print(response.text)
```

```python
from bs4 import BeautifulSoup
import requests

url = "https://www.yes24.com/"
response = requests.get(url)

response.encoding = response.apparent_encoding
soup = BeautifulSoup(response.content, "html.parser") # text - 문자열 데이터, content - 바이너리 데이터
print(soup.title)
print(soup.find("title"))
```

```python
print(soup.find("div").text)
print(soup.find("h2").text)
print(soup.find("li").text)
```

```python
# 예스24 메인 페이지의 실제 컨테이너 ID인 'yesWrap'을 사용합니다.
# 요소가 존재하지 않을 경우를 대비해 예외 처리를 추가합니다.

content = soup.find(id="yesWrap")

if content:
    print(content.prettify()[:1000]) # 내용이 너무 길어질 수 있어 앞부분 1000자만 출력
else:
    print("해당 ID를 가진 요소를 찾을 수 없습니다. 'wrapper' 또는 'yesBody' 등 다른 ID를 시도해보세요.")
```

```python
# 예스24 메인 페이지의 실제 구조에 맞춰 클래스명을 수정합니다.
# 'tp01' 클래스는 메인 페이지의 주요 컨텐츠 섹션을 나타냅니다.

target_section = soup.find(class_="tp01")

if target_section:
    print("--- 'tp01' 클래스 요소 찾기 성공 ---")
    print(target_section.prettify()[:1000])
else:
    # 만약 tp01도 없다면, 헤더 영역(yesHeader)을 찾아봅니다.
    header_section = soup.find(id="yesHeaderArea")
    if header_section:
        print("--- 'yesHeaderArea' ID 요소 찾기 성공 ---")
        print(header_section.prettify()[:1000])
    else:
        print("클래스나 ID를 찾을 수 없습니다. soup.prettify()를 통해 전체 구조를 먼저 확인해보세요.")
```

```python
# 1. 베스트셀러 페이지로 다시 요청을 보냅니다.
best_url = "https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001"
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/121.0.0.0 Safari/537.36'}
res = requests.get(best_url, headers=headers)
soup_best = BeautifulSoup(res.text, 'html.parser')

# 2. 베스트셀러 리스트 영역 찾기
best_list = soup_best.find(id="yesBestList")

if best_list:
    print("--- 예스24 베스트셀러 목록 ---")
    # 3. 리스트 내의 각 도서 요소(li) 찾기
    books = best_list.find_all("li")

    for book in books:
        # 4. 도서 제목이 들어있는 클래스(.gd_name) 찾기
        title_tag = book.find(class_="gd_name")
        if title_tag:
            print(title_tag.text.strip())
else:
    print("베스트셀러 목록을 찾을 수 없습니다. URL이나 페이지 로딩 상태를 확인해주세요.")
```

```python
# 베스트셀러 목록에서 도서 제목과 상세 링크(URL)를 추출합니다.
if best_list:
    books = best_list.find_all("li")
    for book in books:
        title_tag = book.find(class_="gd_name")
        if title_tag:
            title = title_tag.text.strip()
            link = "https://www.yes24.com" + title_tag['href']
            print(f"제목: {title}")
            print(f"링크: {link}")
else:
    print("데이터를 찾을 수 없습니다. 이전 셀(5I4un5BwWoJE)이 먼저 실행되어야 합니다.")
```

```python
import requests
from bs4 import BeautifulSoup
import time

# 예스24 베스트셀러 여러 페이지 크롤링 예시
headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/121.0.0.0 Safari/537.36'}

for page in range(1, 4):  # 1페이지부터 3페이지까지
    print(f"\n--- 현재 {page}페이지 크롤링 중 ---")
    list_url = f'https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001&pageNumber={page}'

    res = requests.get(list_url, headers=headers)
    soup = BeautifulSoup(res.text, "html.parser")

    # 목록에서 각 도서의 링크 찾기
    books = soup.select('#yesBestList > li')

    for book in books:
        title_tag = book.select_one('.gd_name')
        if title_tag:
            title = title_tag.text.strip()
            detail_url = "https://www.yes24.com" + title_tag['href']

            # 상세 페이지로 이동
            detail_res = requests.get(detail_url, headers=headers)
            detail_soup = BeautifulSoup(detail_res.text, "html.parser")

            # 상세 페이지에서 판매지수 정보 추출
            sell_num = detail_soup.select_one('.gd_sellNum')
            sell_text = sell_num.get_text(strip=True) if sell_num else "판매지수 정보 없음"

            print(f"[도서명]: {title}")
            print(f"[판매지수]: {sell_text}")

            # 서버 부하 방지를 위한 짧은 휴식
            time.sleep(0.5)

    time.sleep(1)
```

```python
import requests
from bs4 import BeautifulSoup
import time

headers = {'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/121.0.0.0 Safari/537.36'}

def get_book_detail(url):
    """상세 페이지에서 추가 정보를 가져오는 함수"""
    detail_res = requests.get(url, headers=headers)
    detail_soup = BeautifulSoup(detail_res.text, "html.parser")
    sell_num = detail_soup.select_one('.gd_sellNum')
    return sell_num.get_text(strip=True) if sell_num else "판매지수 없음"

def get_bestseller_list(page):
    """목록 페이지에서 도서 리스트를 가져오는 함수"""
    print(f"\n--- 현재 {page}페이지 수집 중 ---")
    list_url = f'https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001&pageNumber={page}'
    res = requests.get(list_url, headers=headers)
    soup = BeautifulSoup(res.text, "html.parser")

    books = soup.select('#yesBestList > li')
    for book in books:
        title_tag = book.select_one('.gd_name')
        if title_tag:
            title = title_tag.text.strip()
            detail_link = "https://www.yes24.com" + title_tag['href']

            # 상세 정보 함수 호출
            sell_info = get_book_detail(detail_link)

            print(f"[제목]: {title}")
            print(f"[상세]: {sell_info}")
            time.sleep(0.5)

# 1~3페이지 실행
for page in range(1, 4):
    get_bestseller_list(page)
    time.sleep(1)
```
